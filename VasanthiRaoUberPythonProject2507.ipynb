```python
if dp>600
```

```python
return(dp-(dp*0.25))
```

```python
elif 350<dp<600:
```

```python
return(dp-(dp*0.15)
```

```python
elif 100<dp-(dp*0.10)
```

```python
else:
```

```python
return(dp-(dp*0.05)
```
